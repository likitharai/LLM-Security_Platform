```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent

PROCESSED = BASE_DIR / "data" / "processed"
OUTPUT = BASE_DIR / "outputs" / "eda"

OUTPUT.mkdir(parents=True, exist_ok=True)

train = pd.read_csv(PROCESSED / "security_train.csv")
validation = pd.read_csv(PROCESSED / "security_validation.csv")
test = pd.read_csv(PROCESSED / "security_test.csv")
```

```python
print("="*60)

print("TRAIN")

print(train.shape)

print("="*60)

print("VALIDATION")

print(validation.shape)

print("="*60)

print("TEST")

print(test.shape)
```

```python
print(train.info())

print(train.describe(include="all"))
```

```python
import matplotlib.pyplot as plt
import os

missing = train.isnull().sum()

print(missing)

missing.plot(
    kind="bar",
    figsize=(8,4)
)

plt.title("Missing Values")
plt.xlabel("Columns")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "missing_values.png"
    ),
    dpi=300
)

plt.close()   # or plt.show() if you also want to display it
```

```python
duplicates = train["prompt"].duplicated().sum()

print("Duplicate prompts:", duplicates)

print(
    "Duplicate %",
    duplicates/len(train)*100
)
```

```python
attack = train["attack_type"].value_counts()

print(attack)

attack.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Attack Distribution")

plt.ylabel("Count")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "attack_distribution.png"
    )
)

plt.close()
```

```python
source = train["source_dataset"].value_counts()

print(source)

source.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Source Dataset Distribution")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "source_distribution.png"
    )
)

plt.close()
```

```python
severity = train["severity"].value_counts().sort_index()

print(severity)

severity.plot(
    kind="bar"
)

plt.title("Severity Distribution")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "severity_distribution.png"
    )
)

plt.close()
```

```python
safe = train["is_safe"].value_counts()

safe.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.title("Safe vs Unsafe")

plt.savefig(
    os.path.join(
        OUTPUT,
        "safe_vs_unsafe.png"
    )
)

plt.close()
```

```python
phi = train["phi_present"].value_counts()

phi.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.title("PHI Distribution")

plt.savefig(
    os.path.join(
        OUTPUT,
        "phi_distribution.png"
    )
)

plt.close()
```

```python
train["characters"] = train["prompt"].str.len()

train["words"] = train["prompt"].str.split().str.len()

print(train["characters"].describe())

print(train["words"].describe())
```

```python
train["words"].hist(
    bins=30
)

plt.title("Prompt Length")

plt.xlabel("Words")

plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "prompt_length_histogram.png"
    )
)

plt.close()
```

```python
train.boxplot(
    column="words",
    by="attack_type",
    rot=45
)

plt.title("Prompt Length by Attack Type")

plt.suptitle("")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "prompt_length_by_attack.png"
    )
)

plt.close()
```

```python
heatmap = pd.crosstab(
    train["attack_type"],
    train["severity"]
)

print(heatmap)

plt.imshow(
    heatmap,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(heatmap.columns)),
    heatmap.columns
)

plt.yticks(
    range(len(heatmap.index)),
    heatmap.index
)

plt.title("Attack Type vs Severity")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT,
        "attack_vs_severity_heatmap.png"
    )
)

plt.close()
```

```python
from wordcloud import WordCloud
text = " ".join(train["prompt"])

wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color="white"
).generate(text)

plt.imshow(wordcloud)

plt.axis("off")

plt.savefig(
    os.path.join(
        OUTPUT,
        "top_words.png"
    )
)

plt.close()
```

```python
with open(
    os.path.join(
        OUTPUT,
        "summary.txt"
    ),
    "w"
) as f:

    f.write("="*50+"\n")

    f.write("Security Dataset Summary\n")

    f.write("="*50+"\n\n")

    f.write(f"Train Shape : {train.shape}\n")

    f.write(f"Validation Shape : {validation.shape}\n")

    f.write(f"Test Shape : {test.shape}\n\n")

    f.write("Attack Distribution\n")

    f.write(str(train["attack_type"].value_counts()))

    f.write("\n\n")

    f.write("Source Distribution\n")

    f.write(str(train["source_dataset"].value_counts()))

    f.write("\n\n")

    f.write("Severity Distribution\n")

    f.write(str(train["severity"].value_counts()))

    f.write("\n\n")

    f.write("Missing Values\n")

    f.write(str(train.isnull().sum()))

    f.write("\n\n")

    f.write(f"Duplicate Prompts : {duplicates}\n")

    f.write(f"Average Words : {train['words'].mean():.2f}\n")

    f.write(f"Average Characters : {train['characters'].mean():.2f}\n")
```
