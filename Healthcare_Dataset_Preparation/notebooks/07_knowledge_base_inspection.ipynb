```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path(
    r"C:\Users\raich\Desktop\major\Healthcare_Dataset_Preparation"
)

KB_PATH = (
    BASE_DIR
    / "data"
    / "processed"
    / "knowledge_base.csv"
)

print("Knowledge Base:", KB_PATH)
print("Exists:", KB_PATH.exists())
```

```python
kb_df = pd.read_csv(KB_PATH)

print("Shape:", kb_df.shape)

print("\nColumns:")
print(kb_df.columns.tolist())

display(kb_df.head(10))
```

```python
print("Number of rows:", len(kb_df))
print("Number of columns:", len(kb_df.columns))

print("\nData types:")
print(kb_df.dtypes)

print("\nMissing values:")
display(
    kb_df.isnull().sum().sort_values(
        ascending=False
    ).to_frame("missing_count")
)
```

```python
duplicate_count = kb_df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("\nExample duplicates:")
    display(
        kb_df[
            kb_df.duplicated(keep=False)
        ].head(10)
    )
```

```python
for column in kb_df.columns:
    print("\n" + "=" * 60)
    print("COLUMN:", column)
    print("=" * 60)
    
    print("Unique values:", kb_df[column].nunique())
    
    display(
        kb_df[column]
        .dropna()
        .astype(str)
        .head(5)
    )
```

```python
if "attack_type" in kb_df.columns:
    print(
        kb_df["attack_type"]
        .value_counts(dropna=False)
    )
```

```python
if "category" in kb_df.columns:
    print(
        kb_df["category"]
        .value_counts(dropna=False)
    )
```

```python
text_candidates = []

for column in kb_df.columns:
    if kb_df[column].dtype == "object":
        text_candidates.append(column)

print("Text-like columns:")
print(text_candidates)
```

```python
for column in text_candidates:
    
    lengths = (
        kb_df[column]
        .fillna("")
        .astype(str)
        .str.len()
    )
    
    print(
        f"{column}: "
        f"mean={lengths.mean():.1f}, "
        f"min={lengths.min()}, "
        f"max={lengths.max()}"
    )
```

```python
TEXT_COLUMN = "content"
```

```python
TEXT_COLUMN = "response" if "response" in kb_df.columns else "prompt"

text_duplicates = (
    kb_df[TEXT_COLUMN]
    .fillna("")
    .astype(str)
    .duplicated()
    .sum()
)

print("Duplicate text entries:", text_duplicates)
```

```python
print("KNOWLEDGE BASE SUMMARY")
print("=" * 60)

print(f"Rows: {len(kb_df)}")
print(f"Columns: {len(kb_df.columns)}")
print(f"Duplicate rows: {kb_df.duplicated().sum()}")

print("\nMissing values:")
print(
    kb_df.isnull()
    .sum()
    .sum()
)

print("\nColumn names:")
for column in kb_df.columns:
    print(f"- {column}")
```

```python
print("SOURCE DATASET DISTRIBUTION")
print(
    kb_df["source_dataset"]
    .value_counts(dropna=False)
)
```

```python
print("SOURCE DATASET × CONTEXT AVAILABILITY")

context_available = kb_df["context"].notna()

source_context = pd.crosstab(
    kb_df["source_dataset"],
    context_available
)

source_context.columns = [
    "context_missing",
    "context_present"
]

display(source_context)
```

```python
duplicate_prompts = kb_df[
    kb_df["prompt"].duplicated(keep=False)
].sort_values("prompt")

print(
    "Rows involved in duplicate prompts:",
    len(duplicate_prompts)
)

display(
    duplicate_prompts.head(20)
)
```

```python
duplicate_responses = kb_df[
    kb_df["response"].duplicated(keep=False)
].sort_values("response")

print(
    "Rows involved in duplicate responses:",
    len(duplicate_responses)
)

display(
    duplicate_responses.head(20)
)
```

```python
prompt_response_duplicates = kb_df[
    kb_df.duplicated(
        subset=["prompt", "response"],
        keep=False
    )
]

print(
    "Duplicate prompt-response pairs:",
    len(prompt_response_duplicates)
)

display(
    prompt_response_duplicates.head(20)
)
```
